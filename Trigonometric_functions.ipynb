```python
import math
```

```python
angle_degree = 30
angle_radian = angle_degree * math.pi / 180
print("angle_degree: ", angle_radian)
```

```python
math.cos(angle_radian); math.sin(angle_radian); math.tan(angle_radian)
```

직교 좌표를 극좌표로 변환

```python
x,y = 1.0, 2.0
```

```python
r = math.sqrt(x**2 + y**2)
theta_radian = math.atan2(y, x)
print(r, theta_radian)
```
