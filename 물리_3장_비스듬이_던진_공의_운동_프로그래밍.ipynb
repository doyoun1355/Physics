```python
import math

# 초기값 설정
V0 = 100
theta = 37
pi = math.pi
theta_radian = theta * pi / 180
y0 = 140
g = -9.80

# 초기 속도 성분
V0x = V0 * math.cos(theta_radian)
V0y = V0 * math.sin(theta_radian)

# (a)지면에 닿기까지의 시간 계산
rg = 0.5 * g
vq = V0y**2 - 4*rg*y0


t1 = (-V0y + math.sqrt(vq)) / (2*g)
t2 = (-V0y - math.sqrt(vq)) / (2*g)

if t1 > 0:
    tf = t1
else:
    tf = t2

# (b)수평 거리 계산
x = V0x * tf

# (c)속도 성분 계산
Vx = V0x
Vy = V0y + g * tf

# (d)속도의 크기 계산
r = math.sqrt(Vx**2 + Vy**2)

# (e)속도 벡터와 수평 방향이 이루는 각도
angle = math.atan2(Vy, Vx) * 180 / pi

# 결과 출력
print(f"지면에 닿기까지의 시간 ==> {tf:.2f}초")
print(f"수평거리 x ==> {x:.2f} m")
print(f"속도의 크기 ==> {r:.2f} m/s")
print(f"속도 벡터와 수평 방향이 이루는 각의 크기 ==> {angle:.2f}도")
```
