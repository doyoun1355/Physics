```python
import math

g = 9.8
pi = math.pi
theta = 30
theta_radian = theta * pi / 180

#초기 속도 입력
V0 = float(input("V0의 값을 입력하시오 ==> "))

# x, y 방향의 초기 속도 계산
V0x = V0 * math.cos(theta_radian)
V0y = V0 * math.sin(theta_radian)

# 최고점 높이 계산
y = (V0y ** 2) / (2 * g)

# 최고점 도달 시간
t = V0y / g

# 수평 도달 거리
R = (V0 ** 2) * math.sin(2 * theta_radian) / g

# 지면에 닿을 때의 Vy, V 계산
Vy = -V0y
V = math.sqrt(V0x ** 2 + Vy ** 2)

# 결과 출력
print(f"x, y 방향의 초기속도 V0x = {V0x:.2f}, V0y = {V0y:.2f}")
print(f"최고점의 높이 ==> {y:.2f}")
print(f"최고점에 도달하는 시간 ==> {t:.2f}")
print(f"수평으로 이동한 최고 거리 ==> {R:.2f}")
print(f"지면에 닿을 때의 속도 벡터 크기 V ==> {V:.2f}")
```
